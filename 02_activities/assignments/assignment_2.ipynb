input: root = [1,2,3,4,5,5,6]

```python
# Your answer here
from collections import deque
# Definition for a binary tree node.
class TreeNode(object):
  def __init__(self, val = 0, left = None, right = None):
    self.val = val
    self.left = left
    self.right = right

def is_duplicate(root: TreeNode) -> int:
  if not root:
    return -1

  # BFS traversal
  queue = deque([(root, 0)]) # Queue stores (node, depth)
  seen = {}  # Dictionary to track occurrences of values

  while queue:
      node, depth = queue.popleft()
      
      # Check if the value has already been seen
      if node.val in seen:
          return node.val
      
      # Record the value and its depth
      seen[node.val] = depth
      
      # Add children to the queue
      if node.left:
          queue.append((node.left, depth + 1))
      if node.right:
          queue.append((node.right, depth + 1))
  
  # No duplicates
  return -1
```

```python
root = TreeNode(1,
                TreeNode(2,TreeNode(4), TreeNode(5)),
                TreeNode(3, TreeNode(5),TreeNode(6)))
```

```python
print(is_duplicate(root))
```

BFS processes nodes level by level, starting from the root. This ensures the duplicate closest to the root is found first.
The seen dictionary efficiently tracks all previously encountered values. Checking and inserting into the dictionary both take 
$O$(1) on average. As soon as a duplicate is found, the function exits with the duplicate value. This ensures minimal traversal.The queue ensures all nodes are visited in the correct order.

Time Complexity: 
BFS processes each node exactly once and these operation takes $O$(1) each. The function processes each node once and each operation is $O$(1). the total time complexity is $O$(n). 
Space Complexity:
Queue:

The maximum size of the queue is determined by the number of nodes at the widest level of the binary tree.
For a balanced binary tree, this is approximately half the number of nodes at the last level
$O$(n/2), which simplifies to:$O$(n). For a skewed binary tree, the queue can grow to $O$(n)
The seen dictionary stores unique node values.
In the worst case, all $n$ nodes have unique values, so the dictionary size will also be $O$(n)
Total Space Complexity:
The queue and dictionary dominate the space usage, so the total space complexity is: $O$(n)
